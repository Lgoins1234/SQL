```python
#importing data and making sure its all good

import pandas as pd

# Load datasets
test_results = pd.read_csv('test_results.csv')
user_table = pd.read_csv('user_table.csv')

# Preview data
print(test_results.head())
print(user_table.head())

# Check data info
print(test_results.info())
print(user_table.info())
```

```python
#Merge dataset

df = test_results.merge(user_table, on='user_id', how='left')
df.head()
```

```python
#cleaning and preparing merged dataset

df.isnull().sum()
df.duplicated().sum()
df['price'].value_counts()
```

```python
#Step 1: Comparing $39 vs $59 Groups
#Group by price and compute metrics

# 1️⃣ Create a revenue column
# Each converted user generates revenue equal to the price they saw
df['revenue'] = df['converted'] * df['price']

# 2️⃣ Group by price and compute key metrics
summary = df.groupby('price').agg(
    users=('user_id', 'count'),
    conversions=('converted', 'sum'),
    total_revenue=('revenue', 'sum')
).reset_index()

# 3️⃣ Calculate conversion rate and average revenue per user (ARPU)
summary['conversion_rate'] = summary['conversions'] / summary['users']
summary['arpu'] = summary['total_revenue'] / summary['users']

# 4️⃣ Display summary table
summary

#Conversion Rate (CR): How many users bought at each price.
#ARPU (Average Revenue Per User): Which price yields more total revenue per visitor.

```

```python
# Step 1 continued: Statistical Significance Test (T-Test or Proportion Test)
# deciding which price performs better
# (a) comparing conversion rates

from statsmodels.stats.proportion import proportions_ztest

# Count of conversions and total users in each group
conv_counts = summary['conversions'].values
nobs = summary['users'].values

z_stat, p_val = proportions_ztest(conv_counts, nobs)
print(f"Z-stat: {z_stat:.3f}, P-value: {p_val:.4f}")

# If p-value < 0.05, the difference in conversion rates is statistically significant.
```

```python
# (b) comparing revenue per user

from scipy.stats import ttest_ind

group_39 = df[df['price'] == 39]['revenue']
group_59 = df[df['price'] == 59]['revenue']

t_stat, p_val = ttest_ind(group_39, group_59, equal_var=False)
print(f"T-stat: {t_stat:.3f}, P-value: {p_val:.4f}")
```

```python
# Step 2: Behavioral Segmentation (User Insights)
# identify segments that respond differently to the pricing

# Choose relevant segmentation columns that exist in your dataset
segments = ['device', 'operative_system', 'country', 'source']

# Loop through each segment and calculate conversion rates by price
for col in segments:
    temp = df.groupby(['price', col])['converted'].mean().unstack()
    print(f"\nConversion rate by {col}:")
    print(temp.round(3))  # rounded for cleaner display

```

```python
# for visualization purposes
import matplotlib.pyplot as plt

df.groupby(['price', 'device'])['converted'].mean().unstack().plot(kind='bar')
plt.title('Conversion Rate by Device and Price')
plt.ylabel('Conversion Rate')
plt.show()

# Which segment converts better overall
# Which ones are more price-sensitive (bigger drop when price increases)

```

```python
# Step 3: Test duration and stability
df['timestamp'] = pd.to_datetime(df['timestamp'])
daily = df.groupby(['timestamp', 'price'])['converted'].mean().unstack()

daily.plot(title='Daily Conversion Rate by Price')

# Whether results stabilized (lines flatten out)
# Whether there’s enough data to conclude (usually after results stop fluctuating)

```

```python
# In conclusion, Although conversion rate is slightly higher at $59, total revenue per user is also higher.
# Statistical tests show the difference in ARPU is significant (p < 0.05).
# Recommendation: Adopt the $59 price but monitor retention and refund rates.
```
