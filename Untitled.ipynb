```python
#Step 7: Total Survey Response rate by all the Hospitals.

# Goal: Calculate the overall Survey Response Rate Percent across all hospitals in the dataset.
# Why Weighted Average?
# Some hospitals complete more surveys than others.
# So instead of just averaging response percentages (which would treat a hospital with 10 surveys the same as one with 10,000),
# I'll compute a weighted average.
# This gives a more accurate and fair representation of the true overall response rate.
```

```python
# Ensure numeric conversion
df['Survey Response Rate Percent'] = pd.to_numeric(df['Survey Response Rate Percent'], errors='coerce')
df['Number of Completed Surveys'] = pd.to_numeric(df['Number of Completed Surveys'], errors='coerce')

# Drop rows missing either field
valid_data = df.dropna(subset=['Survey Response Rate Percent', 'Number of Completed Surveys'])

# Calculate weighted average response rate
total_surveys = valid_data['Number of Completed Surveys'].sum()
weighted_response_sum = (valid_data['Survey Response Rate Percent'] * valid_data['Number of Completed Surveys']).sum()
overall_response_rate = weighted_response_sum / total_surveys

print(f"Overall Survey Response Rate: {overall_response_rate:.2f}%")

overall_response_rate
```

```python
plt.figure(figsize=(5, 6))
plt.bar(['All Hospitals'], [overall_response_rate])
plt.ylabel('Overall Survey Response Rate (%)')
plt.title('Total Survey Response Rate Across All Hospitals')
plt.text(0, overall_response_rate + 0.2, f"{overall_response_rate:.2f}%", ha='center')
plt.show()
```

```python
# Step 8: Build a report that will give me some new insight from data

# Do hospitals with higher survey response rates also tend to have higher patient survey star ratings?
# This tells me whether patient engagement correlates with hospital quality perception.
# Goal: Explore whether higher patient participation (survey response rates) is associated with higher hospital satisfaction (star rating).
# Why This Matters:
# A high response rate may mean patients feel encouraged and valued.
# Hospitals with low response rates may have patient trust or communication issues.
# Understanding this relationship supports quality improvement and patient outreach strategy.

# Approach:
# Convert necessary columns to numeric.
# Drop missing / invalid values.
# Compute the correlation between:
# Patient Survey Star Rating
# Survey Response Rate Percent
# Create a scatter plot to visually assess the trend.

```

```python
# Convert to numeric
df['Survey Response Rate Percent'] = pd.to_numeric(df['Survey Response Rate Percent'], errors='coerce')
df['Patient Survey Star Rating'] = pd.to_numeric(df['Patient Survey Star Rating'], errors='coerce')

# Remove rows missing data
rating_data = df.dropna(subset=['Survey Response Rate Percent', 'Patient Survey Star Rating'])

# Compute correlation
correlation = rating_data['Survey Response Rate Percent'].corr(rating_data['Patient Survey Star Rating'])
correlation

# If correlation value returns:
# +0.4 or higher → Moderate positive relationship
# 0.1 to 0.3 → Weak but present positive relationship
# Near 0 → No meaningful relationship
# Negative value → Higher response rates linked to lower ratings (uncommon)
```

```python
plt.figure(figsize=(8,6))
plt.scatter(rating_data['Survey Response Rate Percent'], rating_data['Patient Survey Star Rating'])
plt.xlabel('Survey Response Rate (%)')
plt.ylabel('Patient Survey Star Rating')
plt.title('Relationship Between Survey Response Rate and Patient Satisfaction Rating')
plt.tight_layout()
plt.show()
```
