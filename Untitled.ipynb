```python
# I'm starting by cleaning the datasets so i can generate valid insights once they are aligned.
# I am combining:
# PBJ staffing data
# SNF VBP performance scores
# CMS vaccination data
# These three datasets live at different levels (facility-level vs state-level) and use different identifiers.
# Cleaning steps ensure:
# CCNs match across datasets
# Column names are consistent
# Numeric fields are correctly typed
# Merging does not fail or create incorrect duplicates
# Without this foundation, any charts, comparisons, or insights would be wrong.
```

```python
import pandas as pd
import warnings

# Suppress dtype warnings
warnings.simplefilter(action='ignore', category=pd.errors.DtypeWarning)

# Load datasets
pbj = pd.read_csv("PBJ_Daily_Nurse_Staffing_Q2_2024.csv", low_memory=False, dtype=str)
vax = pd.read_csv("NH_CovidVaxAverages_20241027.csv", low_memory=False, dtype=str)
vbp = pd.read_csv("FY_2024_SNF_VBP_Facility_Performance.csv", low_memory=False, dtype=str)

# Inspect
print("PBJ shape:", pbj.shape)
print("Vaccination shape:", vax.shape)
print("VBP shape:", vbp.shape)

print("\nPBJ preview:")
display(pbj.head())

print("\nVaccination preview:")
display(vax.head())

print("\nVBP preview:")
display(vbp.head())
```

```python
import pandas as pd

pbj = pd.read_csv("PBJ_Daily_Nurse_Staffing_Q2_2024.csv", low_memory=False)
vax = pd.read_csv("NH_CovidVaxAverages_20241027.csv", low_memory=False)
snf_vbp = pd.read_csv("FY_2024_SNF_VBP_Facility_Performance.csv", low_memory=False)

print("PBJ loaded:", pbj.shape)
print("Vaccination loaded:", vax.shape)
print("SNF VBP loaded:", snf_vbp.shape)
```

```python
# Step 2: Inspect & Clean SNF VBP Column Names

# Show original columns (already done but included for completeness)
snf_vbp.columns.tolist()

# Create a cleaned version of the dataset
snf = snf_vbp.copy()

# Standardize column names
snf.columns = (
    snf.columns
       .str.lower()
       .str.strip()
       .str.replace(" ", "_")
       .str.replace("(", "", regex=False)
       .str.replace(")", "", regex=False)
       .str.replace("-", "_")
       .str.replace("/", "_")
       .str.replace(":", "")
)

# Remove all footnote columns
snf = snf[[col for col in snf.columns if "footnote" not in col]]

snf.head()



```

```python
snf.columns.tolist()
```

```python
# im fixing the CCN formatting so that it could be used for later merging

# Fix CCN formatting in SNF VBP dataset
snf["ccn"] = (
    snf["cms_certification_number_ccn"]
        .astype(str)
        .str.extract(r"(\d+)", expand=False)  # extract digits only
        .fillna("")
        .str.zfill(6)                        # pad to 6 digits
)

snf["ccn"].head()
```

```python
# Step 3 — Clean PBJ Dataset
# I will clean and zero-pad provnum, convert numeric staffing columns into proper numeric types,
# and aggregate staffing at the facility level

# Step 3: Clean PBJ Dataset

import pandas as pd

pbj_clean = pbj.copy()

# --- 1. Clean CCN (provnum) ---
# Extract digits only and zero-pad to 6 characters
pbj_clean["ccn"] = (
    pbj_clean["PROVNUM"]
        .astype(str)
        .str.extract(r"(\d+)", expand=False)
        .fillna("")
        .str.zfill(6)
)

print("Sample PBJ CCNs:", pbj_clean["ccn"].head(), "\n")

# --- 2. Convert staffing hour columns to numeric ---

staff_cols = [
    col for col in pbj_clean.columns 
    if col.startswith("hrs_")  # all staffing hour columns
]

pbj_clean[staff_cols] = pbj_clean[staff_cols].apply(pd.to_numeric, errors="coerce")

print("Converted staffing columns:", staff_cols, "\n")

# --- 3. Aggregate PBJ staffing by facility (mean hours per quarter) ---

pbj_facility = (
    pbj_clean
        .groupby("ccn")[staff_cols + ["MDScensus"]]
        .mean()
        .reset_index()
)

print("PBJ facility-level dataset shape:", pbj_facility.shape)
pbj_facility.head()

```

```python
# STEP 4 — Merge PBJ + SNF VBP Datasets

# This is the most important preprocessing step, because:

# PBJ = staffing hours (facility-level)

# SNF VBP = performance scores (facility-level)

# Both use CCN as their unique facility ID

# Vaccination dataset has no facility-level CCN, so we will merge it later at the state level
```

```python
# Step 4: Merge PBJ + SNF VBP datasets on CCN

# First ensure CCN exists in both datasets
print("PBJ has CCN:", "ccn" in pbj_clean.columns)
print("SNF has CCN:", "ccn" in snf.columns)

# Merge on CCN
merged = pd.merge(
    pbj_clean,
    snf,
    on="ccn",
    how="inner"   # only include facilities present in both datasets
)

print("Merged dataset shape:", merged.shape)
merged.head()
```

```python
# STEP 5 — Add Vaccination Data (State-Level Merge)
# What I will do:

# Clean vaccination column names
# Convert vaccination percentages to numeric
# Aggregate by state
# Merge state vaccination metrics into your facility-level merged dataset
```

```python
# Step 5: Clean Vaccination Dataset (state-level)

vax_clean = vax.copy()

# Standardize column names
vax_clean.columns = (
    vax_clean.columns
        .str.lower()
        .str.strip()
        .str.replace(" ", "_")
        .str.replace("%", "")
        .str.replace("-", "_")
)

# Convert vaccination percentages to numeric
vax_clean["percent_of_residents_who_are_up_to_date_on_their_vaccines"] = (
    pd.to_numeric(
        vax_clean["percent_of_residents_who_are_up_to_date_on_their_vaccines"],
        errors="coerce"
    )
)

vax_clean["percent_of_staff_who_are_up_to_date_on_their_vaccines"] = (
    pd.to_numeric(
        vax_clean["percent_of_staff_who_are_up_to_date_on_their_vaccines"],
        errors="coerce"
    )
)

# Keep only useful columns
vax_state = vax_clean.groupby("state").agg({
    "percent_of_residents_who_are_up_to_date_on_their_vaccines": "mean",
    "percent_of_staff_who_are_up_to_date_on_their_vaccines": "mean"
}).reset_index()

vax_state.head()
```

```python
# Add state-level vaccination metrics to merged facility-level dataset

# First ensure 'state' exists in merged
print("Does merged dataset have a state column?", "state" in merged.columns)

merged_with_vax = pd.merge(
    merged,
    vax_state,
    on="state",
    how="left"
)

print("Merged-with-vax shape:", merged_with_vax.shape)
merged_with_vax.head()
```

```python
# STEP 6 — Analysis & Visualizations
# In this steps i will make visualizations that'll support my 5 recommendations.
```

```python
# Step 6A: Correlation analysis

import matplotlib.pyplot as plt
import numpy as np

# Step 6A: Correlation analysis (pure matplotlib)

# Compute correlation matrix
numeric_cols = merged_with_vax.select_dtypes(include=['float', 'int']).columns
corr = merged_with_vax[numeric_cols].corr()

# Plot using matplotlib only
plt.figure(figsize=(12, 10))
plt.imshow(corr, cmap="coolwarm", interpolation="nearest")
plt.colorbar()

# Add labels
plt.xticks(ticks=np.arange(len(corr.columns)), labels=corr.columns, rotation=90)
plt.yticks(ticks=np.arange(len(corr.columns)), labels=corr.columns)

plt.title("Correlation Matrix (Staffing, Census, Performance, Vaccination)")
plt.tight_layout()
plt.show()

# High positive correlations (dark red)

# staffing hours ↑ → performance score ↑

# hours per resident ↑ → achievement ↓ or ↑

# census vs staffing numbers

# High negative correlations (dark blue)

# low staffing → low performance

# understaffed → higher readmission risk

# EXPECTED INSIGHTS:

# hrs_rn and performance_score often show the strongest relationship

# CNA hours also correlate with readmission rates

# Census-normalized staffing hours usually predict VBP outcomes

# This correlation table becomes Recommendation #1 and #2.

```

```python
# 6B: Staffing vs Performance Scatterplot

import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))
plt.scatter(
    merged_with_vax["Hrs_RN"], 
    merged_with_vax["performance_score"], 
    alpha=0.4
)

plt.title("RN Hours vs Performance Score")
plt.xlabel("Average RN Hours (PBJ)")
plt.ylabel("SNF VBP Performance Score")
plt.grid(True)
plt.show()

# Facilities with low staffing + low performance cluster in the bottom-left quadrant → 
# these become high-opportunity targets for Clipboard Health.
# This chart supports Recommendation #1.

```

```python
# Step 6c: State Vaccination vs Staffing (Aggregated)
# Shows if certain states have low vaccination & low staffing → a strong compliance-risk story.


state_summary = merged_with_vax.groupby("state").agg({
    "Hrs_RN": "mean",
    "Hrs_CNA": "mean",
    "percent_of_staff_who_are_up_to_date_on_their_vaccines": "mean"
}).reset_index()

state_summary.sort_values("Hrs_RN").head(10)

# Expected insight:
# States with:

# low RN/CNA hours
# low vaccination rates
# are the best for targeted messaging around staffing stabilization + infection prevention.
# This supports Recommendation #3.

```

```python
# Step 6D: High-Opportunity Facility Ranking
# This'll identify facilities that are:

# Understaffed (low RN & CNA hours)
# Underperforming (low VBP score)
# Higher-impact (large census)
# These become the best sales outreach targets for Clipboard Health.

# Step 6D: High-opportunity scoring using correct column names

# Create a risk score:
# lower staffing → higher risk
# lower performance score → higher risk
# higher census → higher risk
merged_with_vax["risk_score"] = (
    merged_with_vax["Hrs_RN"].rank(ascending=True) + 
    merged_with_vax["Hrs_CNA"].rank(ascending=True) + 
    merged_with_vax["performance_score"].rank(ascending=True) + 
    merged_with_vax["MDScensus"].rank(ascending=False)
)

# Top 25 highest-risk (highest-need) facilities
top_opportunities = merged_with_vax.sort_values("risk_score").head(25)

top_opportunities[[
    "ccn", 
    "state", 
    "Hrs_RN", 
    "Hrs_CNA", 
    "performance_score", 
    "MDScensus",
    "risk_score"
]]



# The top 25 facilities become your sales target list:

# underperforming
# understaffed
# high potential to benefit from Clipboard Health’s staffing solutions

#This supports Recommendation #4 and #5.
```
