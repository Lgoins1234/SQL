```python
# Feature Engineering
# Step 1c: Time-based features

# Assume device_daily is your aggregated dataset
device_daily["day_of_week"] = device_daily["date"].dt.day_name()  # Monday, Tuesday, etc.
device_daily["month"] = device_daily["date"].dt.month              # 1-12
device_daily["week_of_year"] = device_daily["date"].dt.isocalendar().week
device_daily["day_of_month"] = device_daily["date"].dt.day
```

```python
# Step 2c: Add lag features

# Sort by device and date first
device_daily = device_daily.sort_values(by=["device_id", "date"])

# Create lag features (previous day demand)
device_daily["demand_lag_1"] = device_daily.groupby("device_id")["demand"].shift(1)
device_daily["demand_lag_7"] = device_daily.groupby("device_id")["demand"].shift(7)  # one-week lag
```

```python
# step 3c: Rolling Statistics

# 7-day rolling average demand
device_daily["demand_7d_avg"] = device_daily.groupby("device_id")["demand"].transform(lambda x: x.rolling(window=7, min_periods=1).mean())

# 14-day rolling sum (for cumulative demand)
device_daily["demand_14d_sum"] = device_daily.groupby("device_id")["demand"].transform(lambda x: x.rolling(window=14, min_periods=1).sum())
```

```python
display(device_daily.head(10))
```

```python
# Importing these features into a machine learning model for forecasting demand.

from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error

# Drop rows with NaN (due to lag features)
ml_data = device_daily.dropna()

# Features (X) and target (y)
X = ml_data[["demand_lag_1", "demand_lag_7", "demand_7d_avg", "month", "day_of_week"]]
X = pd.get_dummies(X, columns=["day_of_week"], drop_first=True)  # encode categorical
y = ml_data["demand"]

# Train/test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

# Train model
rf = RandomForestRegressor(n_estimators=200, random_state=42)
rf.fit(X_train, y_train)

# Predict
y_pred = rf.predict(X_test)

# Evaluate
mae = mean_absolute_error(y_test, y_pred)
print("MAE:", mae)

# Plot
plt.figure(figsize=(10,5))
plt.plot(y_test.index, y_test, label="Actual Demand")
plt.plot(y_test.index, y_pred, label="Predicted Demand (RF)", color="red")
plt.legend()
plt.show()
```

```python
# Step 4d: Check seasonality numerically

# Average demand per month (to detect monthly seasonality)
avg_monthly = device_daily.groupby(device_daily["date"].dt.month)["demand"].mean()
plt.figure(figsize=(8,4))
avg_monthly.plot(kind="bar", color="orange")
plt.title("Average Demand by Month")
plt.xlabel("Month")
plt.ylabel("Average Quantity Dispensed")
plt.grid(True)
plt.show()

# Average demand per weekday (to detect weekly seasonality)
avg_weekday = device_daily.groupby(device_daily["date"].dt.day_name())["demand"].mean()
avg_weekday = avg_weekday.reindex(["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"])

plt.figure(figsize=(8,4))
avg_weekday.plot(kind="bar", color="green")
plt.title("Average Demand by Weekday")
plt.xlabel("Weekday")
plt.ylabel("Average Quantity Dispensed")
plt.grid(True)
plt.show()
```

```python
# Demand statistics per SKU
sku_variability = device_daily_sku.groupby("sku")["demand"].agg(["mean","std"]).reset_index()

# Coefficient of Variation (std / mean)
sku_variability["cv"] = sku_variability["std"] / sku_variability["mean"]

# Top 10 most variable SKUs
high_variability_skus = sku_variability.sort_values("cv", ascending=False)
high_variability_skus.head(10)
```

```python
# Device Utilization
# Step 1g: Calculate basic device utilization

import pandas as pd

# Ensure device_daily is loaded
device_daily.head()
```

```python
# Calculate total demand per device
device_total_demand = device_daily.groupby("device_id")["demand"].sum().reset_index()

# Count number of active days per device
device_active_days = device_daily.groupby("device_id")["date"].nunique().reset_index()
device_active_days.rename(columns={"date": "active_days"}, inplace=True)

# Merge totals and active days
device_utilization = pd.merge(device_total_demand, device_active_days, on="device_id")

# Calculate average daily demand (as a simple utilization metric)
device_utilization["avg_daily_demand"] = device_utilization["demand"] / device_utilization["active_days"]

# Preview
device_utilization.sort_values("avg_daily_demand", ascending=False).head(10)
```

```python
# Step 2g: visualize device utilization

import matplotlib.pyplot as plt

# Top 10 devices by average daily demand
top_devices = device_utilization.sort_values("avg_daily_demand", ascending=False).head(10)

plt.figure(figsize=(12,6))
plt.bar(top_devices["device_id"], top_devices["avg_daily_demand"], color="teal")
plt.title("Top 10 Devices by Average Daily Demand")
plt.xlabel("Device ID")
plt.ylabel("Average Daily Demand")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```
