```python
#loading and exploring data

import pandas as pd

# Load dataset
df = pd.read_csv("referral.csv")

# Quick look
df.head()
df.info()
df.describe()
```

```python
# Data preparation
# clean the data and prepare for analysis
# I'm preparing the data by creating a period column to separate 
# Before vs After the referral program launch and a converted column to indicate if the user made a purchase.


# Convert date to datetime
df['date'] = pd.to_datetime(df['date'])

# Flag users as Before/After program launch
program_start = pd.to_datetime("2024-10-31")
df['period'] = df['date'].apply(lambda x: 'Before' if x < program_start else 'After')

# Create conversion flag (True if user made a purchase)
df['converted'] = df['money_spent'] > 0

# Quick check
df.head()
```

```python
# Step 1: User Growth Analysis
# compare the number of new users before and after the referral program launch to see if the program increased signups.

# Step 4: User Growth Analysis without plots
user_growth = df.groupby('period')['user_id'].nunique()
print("User Growth Before vs After Referral Program:")
print(user_growth)
```

```python
# optional: create a table showing % increase

before = user_growth.get('Before', 0)
after = user_growth.get('After', 0)
increase_pct = ((after - before) / before) * 100 if before > 0 else 0
print(f"Percentage increase in users after program launch: {increase_pct:.2f}%")
```

```python
# Step 2: Referral conversion Analysis
# calculate and compare the conversion rates of referred and non-referred users. 
# This helps us understand if referrals lead to more purchases.

# Conversion rate for referred vs non-referred users
referral_conversion = df[df['is_referral'] == True]['converted'].mean()
non_referral_conversion = df[df['is_referral'] == False]['converted'].mean()

print("Conversion Rates:")
print(f"Referral Users: {referral_conversion:.2%}")
print(f"Non-Referral Users: {non_referral_conversion:.2%}")

# Optional: Table format
conversion_table = pd.DataFrame({
    'User Type': ['Referral', 'Non-Referral'],
    'Conversion Rate': [referral_conversion, non_referral_conversion]
})
print("\nConversion Table:")
print(conversion_table)

```

```python
# Step 3: Transaction Volume & Revenue Analysis
# assess the financial impact of referrals by calculating total revenue from referred users 
# and their average spend compared to non-referred users.

# Revenue from referred users
referral_revenue = df[df['is_referral'] == True]['money_spent'].sum()
total_revenue = df['money_spent'].sum()
referral_revenue_share = referral_revenue / total_revenue if total_revenue > 0 else 0

print("Revenue Analysis:")
print(f"Revenue from referred users: ${referral_revenue:.2f}")
print(f"Total revenue: ${total_revenue:.2f}")
print(f"Referral revenue share: {referral_revenue_share:.2%}")

# Average spend per user
avg_spent = df.groupby('is_referral')['money_spent'].mean()
print("\nAverage Money Spent per User:")
print(avg_spent)
```

```python
# Step 4a: Descriptive Statistics by Period
# examines user spending before and after the referral program launch. 
# "After" group has no data so 
#Even if we have a small sample, descriptive statistics like count, mean, median, min, and max 
# provide insight into how spending behavior may have changed.

# Step 4a: Descriptive Statistics by Period

# Separate money_spent by period
before_money = df[df['period'] == 'Before']['money_spent']
after_money = df[df['period'] == 'After']['money_spent']

# Check sample sizes
before_count = len(before_money)
after_count = len(after_money)

print(f"Number of users Before: {before_count}")
print(f"Number of users After: {after_count}\n")

# Descriptive stats
if before_count > 0:
    print("Spending Stats - Before Program:")
    print(before_money.describe())
else:
    print("No 'Before' data available.")

if after_count > 0:
    print("\nSpending Stats - After Program:")
    print(after_money.describe())
else:
    print("No 'After' data available.")

```

```python
# Step 4b: Referral vs Non-Referral Comparisons
# compares spending between referred and non-referred users

# Step 4b: Spending Analysis by Referral Status

# Separate money_spent by referral status
referral_money = df[df['is_referral'] == True]['money_spent']
non_referral_money = df[df['is_referral'] == False]['money_spent']

print("Spending Stats - Referred Users:")
print(referral_money.describe())

print("\nSpending Stats - Non-Referred Users:")
print(non_referral_money.describe())

# Percent difference in average spend
if len(referral_money) > 0 and len(non_referral_money) > 0:
    referral_avg = referral_money.mean()
    non_referral_avg = non_referral_money.mean()
    pct_diff = ((referral_avg - non_referral_avg) / non_referral_avg) * 100 if non_referral_avg != 0 else 0
    print(f"\nReferred users spend {pct_diff:.2f}% more than non-referred users on average.")
```

```python
# Step 5: Segmentation Analysis
# Segmenting by country or device can identify high-performing markets or platforms, helping optimize future referral campaigns.

# Conversion by country
country_conversion = df.groupby('country')['converted'].mean().sort_values(ascending=False)
print("Top Countries by Conversion Rate:")
print(country_conversion.head(10))

# Conversion by device (optional)
device_conversion = df.groupby('device_id')['converted'].mean().sort_values(ascending=False)
print("\nTop Devices by Conversion Rate:")
print(device_conversion.head(10))
```
